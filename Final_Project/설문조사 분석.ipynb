```python
import pandas as pd
import matplotlib.pyplot as plt

# 한글 깨짐 방지
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# CSV 불러오기
df = pd.read_csv("지역 기반 혜택·정보 공유 서비스 「기특기특(가칭)」 설문조사.csv", encoding="utf-8-sig")

# 데이터 확인
df.head()
```

```python
print(f"총 응답자 수 : {len(df)}명")
```

```python
df.columns = df.columns.str.strip()
```

```python
df.columns = df.columns.str.strip()
```

```python
target = df[
    (df['Q9. 평소 생활비를 절약하기 위해 노력하는 편인가요?'].isin(['그렇다', '매우 그렇다'])) &
    (df['Q12. 생활비 절약 정보를 한 곳에서 확인할 수 있는 서비스가 있다면 이용할 의향이 있으신가요?'].isin(['있다', '매우 있다']))
]
```

```python
print(f"전체 응답자 : {len(df)}명")
print(f"핵심 타겟 : {len(target)}명")
print(f"비율 : {round(len(target)/len(df)*100,1)}%")
```

```python
age_analysis = target['Q1. 연령대를 선택해주세요.'].value_counts()

age_df = pd.DataFrame({
    '응답수': age_analysis,
    '비율(%)': round(age_analysis / len(target) * 100, 1)
})

age_df
```

```python
# 생활비 절약 의지가 높고(Q9), 서비스 이용 의향도 높은(Q12) 핵심 관심 그룹 추출
high_interest_group = df[
    (df['Q9. 평소 생활비를 절약하기 위해 노력하는 편인가요?'].isin(['그렇다', '매우 그렇다'])) &
    (df['Q12. 생활비 절약 정보를 한 곳에서 확인할 수 있는 서비스가 있다면 이용할 의향이 있으신가요?'].isin(['있다', '매우 있다']))
]

# 그룹 규모 확인
print(f'핵심 관심 그룹: {len(high_interest_group)}명')
print(f'전체 대비 비율: {round(len(high_interest_group)/len(df)*100,1)}%')
print()

# 연령 분포
age_result = high_interest_group['Q1. 연령대를 선택해주세요.'].value_counts()
age_table = pd.DataFrame({
    '응답수': age_result,
    '비율(%)': round(age_result / len(high_interest_group) * 100, 1)
})

print('===== 연령 분포 =====')
print(age_table)
print()

# 소득 분포
income_result = high_interest_group['Q4. 월 평균 소득을 선택해주세요.'].value_counts()
income_table = pd.DataFrame({
    '응답수': income_result,
    '비율(%)': round(income_result / len(high_interest_group) * 100, 1)
})

print('===== 소득 분포 =====')
print(income_table)
print()

# 연령 × 소득 교차분석
print('===== 연령 × 소득 교차분석 =====')
cross_table = pd.crosstab(
    high_interest_group['Q1. 연령대를 선택해주세요.'],
    high_interest_group['Q4. 월 평균 소득을 선택해주세요.']
)

print(cross_table)


import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
age_result.plot(kind='bar')
plt.title('핵심 관심 그룹 연령 분포')
plt.xlabel('연령대')
plt.ylabel('응답 수')
plt.xticks(rotation=0)
plt.show()

plt.figure(figsize=(8,4))
income_result.plot(kind='bar')
plt.title('핵심 관심 그룹 소득 분포')
plt.xlabel('소득 구간')
plt.ylabel('응답 수')
plt.xticks(rotation=30)
plt.show()
```

```python
score_map_q9 = {
    '전혀 그렇지 않다': 1,
    '그렇지 않다': 2,
    '보통이다': 3,
    '그렇다': 4,
    '매우 그렇다': 5
}

score_map_q12 = {
    '전혀 없다': 1,
    '없다': 2,
    '보통이다': 3,
    '있다': 4,
    '매우 있다': 5
}

df['Q9_score'] = df['Q9. 평소 생활비를 절약하기 위해 노력하는 편인가요?'].map(score_map_q9)
df['Q12_score'] = df['Q12. 생활비 절약 정보를 한 곳에서 확인할 수 있는 서비스가 있다면 이용할 의향이 있으신가요?'].map(score_map_q12)

# 두 점수의 평균 = 핵심 관심도
df['target_score'] = (df['Q9_score'] + df['Q12_score']) / 2

# 연령별 평균 관심도
age_score = df.groupby('Q1. 연령대를 선택해주세요.')['target_score'].agg(['count', 'mean']).round(2)
age_score = age_score.sort_values('mean', ascending=False)

print('===== 연령별 핵심 관심도 =====')
print(age_score)
print()

# 소득별 평균 관심도
income_score = df.groupby('Q4. 월 평균 소득을 선택해주세요.')['target_score'].agg(['count', 'mean']).round(2)
income_score = income_score.sort_values('mean', ascending=False)

print('===== 소득별 핵심 관심도 =====')
print(income_score)
print()

# 연령 × 소득별 평균 관심도
cross_score = pd.pivot_table(
    df,
    values='target_score',
    index='Q1. 연령대를 선택해주세요.',
    columns='Q4. 월 평균 소득을 선택해주세요.',
    aggfunc='mean'
).round(2)

print('===== 연령 × 소득별 평균 관심도 =====')
print(cross_score)
```
