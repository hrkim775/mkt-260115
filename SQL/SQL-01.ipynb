데이터 저장
- 데이터는 무엇인가?
- 저장공간은 어떻게 이루어져있나?
- 저장 시, 유의사항이 뭐가 있을까?

데이터 저장을 위해 알고있어야하는 중요한 개념
1. 데이터를 정의 할 수 있어야한다
- 관찰, 측정, 조사 등을 통해 수집된 사실, 값, 기록 등의 원초적인 정보
- 데이터는 그 자체로는 의미부여가 많이 부족함
- 데이터가 어떠한 가치를 가지려면, 쌓여지고 축척된 데이터가 하나의 구조화가 되어야 비로서 의미있는 가치를 부여할 수 있다

2. 데이터가 구조화 된다는 것은 데이터베이스를 만들어 낸다는 것이다 (Database = DB)
- 여러가지의 다양한 데이터를 구조적이고 체계적으로 저장 및 관리 할 수 있는 저장소
- 2가지의 종류로 나뉜다
- 1) 관계형 데이터베이스 : RDBMS -> 구조화 되어있는 다양한 데이터들이 서로 관계를 맺고 있는 형태를 의미
> 오랜 역사를 가지고 있다
> 대기업, 글로벌 기업은 관계형 데이터베이스를 사용
> 안정적, 검증된 데이터 관리 시스템
> 약점 : 서로 다른 데이터가 관계를 맺기 위해서는 기본적으로 사용해야하는 문법 체계가 있다 -> 문법의 의존도가 높음 / 문법을 잘 알지 못하면 사용하기 어렵다
> 데이터가 우선? VS 데이터를 다루는 언어의 문법 체계가 우선? / 이런 생각을 하게 됨
> Oracle, MySQL


- 2) 비관계형 데이터베이스 : CRUD -> 구조화 되어있는 데이터들의 관계에 너무 치중하지 않고 데이터의 입출력 및 수정, 편집 등을 원활하게 하기 위한 목적
> Create / Read / Update / Delete
> 스타트업, 데이터의 입출력이 빈번하게 일어나는 글로벌 기업이 사용
> MongoDB
> NoSQL => Not Only SQL
> 문법체계가 유연하다 = 데이터에 포커스가 맞춰져있다

3. 스키마 (Schema)
- 데이터를 체계적으로 구조화한 것이 데이터베이스라고 한다 그러면 데이터베이스를 생성했다는 것은 어떤 규칙과 패턴을 가지고 각 데이터들을 체계적으로 구조화했다는 뜻
- 데이터를 데이터베이스화 하기 위한 목적으로 어떤 약속된 구조(틀)를 정의해놓은 것을 스키마라고 한다
- 전체 몇개의 칼럼을 기준으로 구조화 했나?
- 각 칼럼 안에 입력되어야할 데이터 값이 어떤 타입?
- 데이터의 타입을 세분화해서 메모리를 얼만큼 가용하게 할것인가?

4. 테이블 (Table)
- 데이터 10만개, 100만개가 있는데 모두 다 독립적이고 개별적으로 사용한다면 비효율적이다
- 비슷한 유형의 데이터를 나눠서 관리하는 것을 테이블이라고 한다
- 데이터를 구조화하기 위해서 데이터베이스 관리 시스템을 적용해야하는데, 이때 다양한 데이터를 유사한 속성에 따라서 행, 열 기반의 테이블로 그룹화할 수 있고 이 것을 데이터베이스 안에 그룹별로 저장한다
- 실질적인 데이터의 저장 단위
- 하나의 데이터베이스 안에는 복수의(여러개의) 테이블이 존재할 수 있다

5. SQL (Strustured Query Language)
- 관계형 데이터베이스를 다루기 위한 문법체계
- DDL (Data Definition Language) : 데이터를 생성, 삭제, 교체
- DML (Date Manipulation Language) : 데이터를 조회, 삽입, 업데이트
- DCL (Date Control Language) : 데이터의 권한, 박탈, 등급화

6. MySQL
- CLI / GUI
- CLI : 명령프롬프트 기반 실행
- GUI : 워크벤치 기반 실행

* 테이블 생성 시, 꼭 이해해야하는 개념
- 데이터베이스를 생성 -> 테이블을 생성 -> 데이터 삽입 및 저장
- 좋은 테이블을 생성했다는 것은 데이터를 최대한 효율적으로 저장했다는 것을 의미
- 최대한 중복되어지는 값을 피한다 -> 테이블 생성 시, 각각의 로우 | 레코드를 식별할 수 있는 고유 키를 생성 = Primary Key (프라이머리 키 = 기본키)
- 다른 외부 데이터와의 호환성 = 관계성이 쉬워야한다 = Foreign Key (포린 키 = 외부키)
