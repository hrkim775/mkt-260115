# SQL (Structured Query Language) 문법

- DDL : Data Definition Language
> CREATE, ALTER, DROP

- DML (데이터 처리 언어) : Date Manipulation Language // 가장 중요하다 (방대하고 복잡)
> INSERT, SELECTOR, GROUP, ORDER
> 초급, 중급, 고급의 차이
> 데이터 타입의 이해도를 본다 : Null, Unknown, 가상쿼리문 & 복합쿼리문 = 서브쿼리
> Create (생성)
> Read (읽기)
> Updata (수정)
> Delete (삭제)

- DCL (데이터 제어 언어) : Data Control Language
> 계정을 생성했는데 이걸 변경하거나 서브계정을 만들거나 팀이랑 함께 사용하고 싶다면 계정을 컨트롤 할 수 있어야한다
> MySQL Cloud 

MySQL, OracleSQL, PostgreSQL은 같은 SQL이지만 소소하게 문법이 다르다

<>과 !=의 차이점
1. <> : 표준SQL문법 => 어떤 SQL이던 다 사용할 수 있다
2. != : MySQL문법
