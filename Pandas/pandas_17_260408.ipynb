```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

df = pd.DataFrame(np.random.rand(10, 2), columns=["A", "B"])

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df.index,
        y=df["A"],
        name="A",
        text=df["A"],
        textposition="auto",
        texttemplate="%{y:.2f}" # 소수점 표시 / 소수점 두번째까지만 표시
    )
)

fig.add_trace(
    go.Bar(
        x=df.index,
        y=df["B"],
        name="B",
        text=df["B"],
        textposition="auto",
        texttemplate="%{y:.2f}"
    )
)

fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

df = pd.DataFrame({
    "A": np.random.randint(10, 50, size=10), # 0~9사이의 정수를 난수값으로 생성한다 / 10부터 50까지의 숫자 10개
    "B": np.random.randint(10, 50, size=10)
 })

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df.index,
        y=df["A"],
        name="A",
        text=df["A"],
        textposition="auto",
        texttemplate="%{y:.2f}" # 정수에서도 사용은 가능하다
    )
)

fig.add_trace(
    go.Bar(
        x=df.index,
        y=df["B"],
        name="B",
        text=df["B"],
        textposition="auto"
    )
    
)

fig.update_layout({
    "title": {
        "text": "Graph with <b>go.Bar</b>", # <b></b> 글씨체 볼드효과
        "x": 0.5,
        "y": 0.9,
        "font": {
            "size": 40
        }
    },
    "showlegend": True,
    "xaxis": {
        "title": "index number",
        "showticklabels": True,
        "dtick": 1
    },
    "yaxis": {
        "title": "randint"
    },
    "autosize": False,
    "width": 880,
    "height": 400
})

fig.show()
```
