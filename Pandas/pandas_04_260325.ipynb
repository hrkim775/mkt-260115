```python
# json
# 자바스크립트의 객체는 파이썬의 딕셔너리와 거의 유사하다
# 다만, 자바스크립트 객체의 경우 {}로 시작하지만 앞에 나오는 객체에 ""를 쓰지 않는다 {name: "David"}
# json  {"name": "David"}
```

```python
# json 형태의 파일을 문자열의 형태로 찾아오는 방법
import json

data = '{"id": "01", "language": "python", "edition": "third", "author": "David Park"}'

json_data = json.loads(data)
print(json_data["id"], json_data["language"], json_data["edition"])
print(type(json_data))
```

```python
# json 형태의 파일을 실제 본 쉐입 형태를 유지한 채로 찾아오는 방법
import json

data = {"id": "01", "language": "python", "edition": "third", "author": "David Park"}

jsondata = json.dumps(data)
print(jsondata, type(jsondata))
```
