데이터 수정 시 유의사항

1. 전체 데이터의 사이즈, 결측치 정보 확인 -> size, shape, info, count
2. 결측치 발생시 대처방안 -> fillna(), dropna()
3. 특정 조건에 해당되는 값 찾기 -> doc[doc.index == "US"]
4. 특정컬럼만 값 찾아오기 -> covid_stat = doc[["Confirmed", "Deaths",	"Recovered"]]
5. 타입 변경시 결측치부터 해결 -> doc2 = doc2.astype({"Confirmed": "int64"})
6. 서로 다른 데이터 연결시, 동일한 컬럼 필드명 통일 -> doc3.columns = ["Country_Region", "Confirmed"]
7. 중복값 존재시 삭제 -> doc4 = doc4.drop_duplicates(subset="Country_Region")

```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "customer_id": [1, 2, 3],
    "customer_name": ["Robert", "Peter", "Dave"]
})
```

```python
df1
```

```python
import pandas as pd

df2 = pd.DataFrame({
    "id": [1, 2, 4],
    "order_id": [100, 200, 300],
    "order_date": ["2021-01-21", "2021-02-03", "2020-10-01"],
    "customer_name": ["Robert", "Peter", "Dave"]
})
```

```python
df2
```

```python
# concat() : 두 개의 데이터프레임을 연결해서 하나의 데이터프레임으로 만드는 방법
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis = 1) # axis는 축을 의미하고 옵션 설정을 하지 않으면 0을 뜻한다. 0은 default값으로 위, 아래로 연결, 1은 좌,우로 연결
```

```python
pd.merge(df1, df2)

# merge() : 공통분모가 되는 것을 기준으로 합하기 때문에 중복되는 값이 나오지 않는다
# concat() : 중복값이 나온다
```
