```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "customer_id": [1, 2, 3],
    "customer_name": ["Robert", "Peter", "David"]
})

df1
```

```python
import pandas as pd

df2 = pd.DataFrame({
    "id" : [1, 2, 4],
    "order_id" : [100, 200, 300],
    "otder_date" : ["2021-01-21", "2021-02-03", "2020-10-01"]
})

df2
```

```python
# concat() : 서로 다른 두 개의 데이터 프레임을 물리적으로 합쳐주는 매서드 (디폴트는 상, 하의 위치로 합친디)

pd.concat([df1, df2])
```

```python
# 0 -> 상하의 위치, 1 -> 위, 아래의 위치

pd.concat([df1, df2], axis=1)
```

```python
# merge() : 서로 다른 두 개의 데이터프레임 안에 동일한 이름을 가진 칼럼을 기준으로 합침

pd.merge(df1, df2) # 각 데이터프레임의 필드명을 왼쪽부터 조회 ->  상호동일한 필드명이 조회되면 그 것을 기준으로 합친다
```

```python
pd.merge(df1, df2, on="id")
```

```python
# merge(데이터프레임1, 데이터프레임2, 연결필드, how=결합방법)
# inner : SQL 이너조인과 동일한 방법으로 연결
# outer : 각 데이터프레임의 인덱스값을 기준으로 서로 다른 데이터프레임을 연결하겠다는 의미
# left : SQL LEFT OUTER JOIN과 동일 (NULL값이 있기 떄문에 잘 사용하지 않는다)
# right : SQL RIGHT OUTER JOIN과 동일 (NULL값이 있기 떄문에 잘 사용하지 않는다)
```

```python
import pandas as pd

pd.merge(df1, df2, on="id", how="inner") # 애초에 디폴트 값이 inner join
```

```python
import pandas as pd

pd.merge(df1, df2, on="id", how="outer")
```

```python
import pandas as pd

pd.merge(df1, df2, on="id", how="left")
```

```python
import pandas as pd

pd.merge(df1, df2, on="id", how="right")
```
