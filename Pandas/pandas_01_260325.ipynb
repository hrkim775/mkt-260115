```python
# 데이터 구조화, 첫 단계는 다양한 형태의 파일을 로드하고 그 파일을 어떻게 편집, 저장할 것인가에서 시작한다
```

```python
# 파일의 정보값을 가져오는 것

data_file = open("00_data/text_data.txt", "r", encoding="utf-8-sig") # ~에 있는 파일을 ~형식으로 인코딩해서 읽어줘
print(data_file)
data_file.close()
```

```python
# with 절은 무언가를 실행한 후, 해당 행위가 모두 종료되면 알아서 기존에 사용하고 있는 메모리 공간 반납, 종료
# 안하면 메모리 누수가 생김

with open("00_data/text_data.txt", "r", encoding="utf-8-sig") as file_desc :
    print(file_desc)
```

```python
# 파일 내용을 읽어오는 것

data_file = open("00_data/text_data.txt", "r", encoding="utf-8-sig")
data_lines = data_file.readlines()
data_lines
```

```python
for data_line in data_lines :
    print(data_line)
    
data_file.close()
```

```python
# Plain Text : 특정 파일을 불러오거나, 작성하거나, 편집/수정하는 작업
# r (read) : read(), readline(), readlines()
# w (write) : overwrite 기능 포함 : 처음부터 문서를 만든다
# a (add | append) : 만들어진 문서 안에 값을 추가한다
```

```python
data_file = open("00_data/text_data_write.txt", "w", encoding="utf-8-sig")
data_file.write("안녕하세요")
data_file.write("David Park 입니다")
data_file.close()
```

```python
data_file = open("00_data/text_data_write.txt", "w", encoding="utf-8-sig")
data_file.write("안녕하세요\n")
data_file.write("David Park 입니다\n")
data_file.close()
```

```python
data_file = open("00_data/text_data_write.txt", "a", encoding="utf-8-sig")
data_file.write("해당 파일은 임시 데이터 입니다")
data_file.close()
```
