```python
# rename() : 이미 연결된 특정 데이터 프레임 내 특정 컬럼명만 수정, 변경 하고자 할때 사용가능한 매서드
# grouping을 할 때, aggregate() | agg() -> 좌측 매서드를 활용해서 연산을 처리할 수도 있음
# pandas -> groupby().sum() | count() -> 특정 요소의 그룹화 진행 -> 그룹 외 요소들의 연산집계 단일화된 연산처리
# 복수의 연산처리 -> 특정 요소의 합도 구하고, 최소값고 구하고 싶음
```

```python
# aggregate() => 집계하다

import pandas as pd

df = pd.DataFrame({
    "미국": [2.1, 2.2, 2.3],
    "한국": [0.4, 0.5, 0.45],
    "중국": [10, 13, 15],
})

df
```

```python
df.aggregate(["sum", "min"]) # 복수연산 가능. agg()와 동일함
```

```python
df.agg(["sum", "min"])
```

```python
df.agg({"미국": "count", "한국": "sum"}) # 필요한 열만 가져와서 독자적인 연산도 가능하고, 복수의 연산도 처리 가능
```

```python
#df.agg("mean", axis="columns") # columns => 열 => 행 단위로 평균값을 집계한다 => 1
df.agg("mean", axis=1)
```

```python
df.agg("mean", axis=0)
```

```python
import pandas as pd

df = pd.DataFrame({
    "년도": [2000, 2010, 2020, 2000, 2000, 2020],
    "미국": [2.1, 2.2, 2.3, 3.1, 4.1, 5.1],
    "한국": [0.4, 0.5, 0.45, 0.7, 0.9, 1.2],
    "중국": [10, 13, 15, 12, 45, 56],
})

df
```

```python
df.groupby("년도").sum()
```

```python
df.groupby("년도").agg({"미국": sum})
```

```python
df.groupby("년도").agg({"미국": "sum", "한국": "mean"})
```

```python
products_category_order = products_category_order.groupby("product_category_name_english").agg({"order_id": "count"}).rename(columns={"order_id": "order_count"})
```

```python
products_category_order
```

```python
products_category_order = products_category_order.sort_values(by="order_count", ascending=False)
products_category_order.head()
```

```python
import plotly.express as px

fig = px.bar(
    x = products_category_order.index,
    y = products_category_order["order_count"],
    title = "Order Count By Product Category",
    labels= {
        "x": "Product Category",
        "y": "Order Count"
    }
)
fig.update_layout(
    xaxis_tickangle=-45
)
fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=products_category_order.index,
        y=products_category_order["order_count"]
    )
)

fig.update_layout(
    {
        "title": {
            "text": "<b>The Number of Order per Category in Brazilian Olist E-Commerce Company</b>",
            "x": 0.5,
            "y": 0.9,
            "font": {
                "size": 16
            }
        },
        "xaxis": {
            "title": "From Oct. 2016 To Sep 2018",
            "showticklabels": True,
            "tickfont": {
                "size": 8
            }
        },
        "yaxis": {
            "title": "The Number of Order"
        },
        "template": "plotly_dark"
    }
)

fig.show()
```
