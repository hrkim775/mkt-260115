```python
# text : plain text
# csv : comma separated values (각각의 행과 열 안에 있는 값을 ,로 구분 짓는다)
"""
name, age
david, 20
jane, 30
"""
```

```python
import csv

data_file = open("00_data/USvideos.csv", "r", encoding="utf-8-sig")
data_lines = csv.reader(data_file, delimiter=",")

for data_line in data_lines :
    print(data_line)
    break

data_file.close()
```

```python
# delimiter : 구분자. 어떤 값이 나오면 해당 문자열을 구분 짓게 할 것인가에 대한 정의
# 기본적으로 csv는 ,로 구분이 되어있기 때문에 구분자를 따로 주지 않아도 ,로 구분되어짐
```

```python
with open("00_data/USvideos.csv", "r", encoding="utf-8-sig") as reader_csv :
    reader = csv.reader(reader_csv, delimiter=",")

    for row in reader :
        print(row)
        break
```

```python
import csv

data_file = open("00_data/tmp_csv.csv", "w", encoding="utf-8-sig")
data_write = csv.writer(data_file, delimiter=",")
data_write.writerow(["1", "2", "3"])
data_file.close()
```

```python
# csv 파일을 리스트 형태로 값을 삽입하는 방법

import csv

with open("00_data/tmp_csv.csv", "w", encoding="utf-8-sig") as writer_csv : 
    writer = csv.writer(writer_csv, delimiter=",")
    writer.writerow(["love"] * 3 + ["banana"])
    writer.writerow(["apple", 2])
    writer.writerow(["dog", "cat", "mouse"])
```

```python
# csv 파일을 딕셔러니 형태로 값을 삽입하는 방법

import csv
with open("00_data/tmp_csv.csv", "w", encoding="utf-8-sig") as writer_csv :
    filed_name_list = ["First_Name", "Last_Name"]
    writer = csv.DictWriter(writer_csv, fieldnames=filed_name_list)
    writer.writeheader()
    writer.writerow({"First_Name": "David", "Last_Name": "Park"})
    writer.writerow({"First_Name": "Dave", "Last_Name": "lee"})
    writer.writerow({"First_Name": "Robin", "Last_Name": "Hood"})
```

```python
import csv

with open ("00_data/tmp_csv.csv", "r", encoding="utf-8-sig") as reader_csv :
    reader = csv.DictReader(reader_csv)
    for row in reader :
        print(row["First_Name"], row["Last_Name"])
```
