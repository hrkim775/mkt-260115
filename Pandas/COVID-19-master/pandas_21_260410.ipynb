# 데이터가 어떤 분야(산업)에서 나온 데이터인지를 확인하고 그에 맞는 분석을 해야한다

- 이커머스 -> 온라인 비즈니스의 한 종류 (고객이 온라인을 통해서 상품을 조회, 구매, 결제를 하고 배송단계를 거쳐 고객에서 상품이 도착하고 고객은 리뷰와 평점을 작성하고 재구매까지 이어진다)
- olist : 비즈니스 중계 플랫폼 형태

- 현재 해당 서비스를 사용하고 있는 고객은 얼마나 있는다? 
- 고객은 주로 어디에서 거주하고 있는가?
- 고객은 주로 어떤 결제 방법을 사용하고 있는가?
- 고객들의 평균 거래액은 얼마일까?
- 일별, 주별, 월별 판매 트랜드는 어떻게 되고있을까?
- 어떤 카테고리에서 많은 매출이 나오고 있는가? -> 어떤 카테고리를 더 성장시켜야 할까? (잘 하는 것을 더 잘하도록-전문성)

- 특정 지역, 국가, 문화적 특징 : 남미쪽은 머리카락에 영혼이 깃들어있다고 생각함 -> 헤어케어 시장이 매우 발달 되어있음

# 데이터 사이즈 = 크기 확인

# 데이터 요소, 속성, 특징 확인 = 수치 (변이, 위치), 범주 (순서, 목록), 관계

1. 수치형 : 5 numners summary (평균, 표준편차, 분산, 최소, 최대, 사분위, 이상치 등)
- 시계열 데이터 차트 : 추이, 트랜드 확인
- 박스플롯, 히스토그램, 산점도, 히트맵 등 데이터에 맞는 차트를 선택

2. 범주형 : 상대빈도, 절대빈도 (각각의 독립적인 데이터 수치 확인을 한다)
- 상대빈도 : 원형차트, 폭포차트, 리본차트 등
- 절대빈도 : 세로, 가로 막대 그래프
