```python
# pandas -> 행과 열을 기반으로 되어있는 테이블형 데이터를 취급하는 전문 라이브러리 모듈
# 데이터를 꼭 테이블 형태로 취급해야하는가?
# 데이터를 취급하고 분석하는 주체는 사람인데, 사람에게 가장 익숙한 형태의 구조여야 잘 쓸 수 있다 그래서 테이블 형태로 취급한다
# pandas는 행, 열을 다루는데 있어서 굉장히 뛰어나기 때문에 이거 제일 많이 사용한다
# 기본적인 row data(원본데이터)를 행, 열을 기반ㄴ으로 변환하고 정체해서 전처리가 가능하다
# 어떤 데이터를 생성할때, 오타, 결측치, 행/열의 위치가 바뀌어있거나 필드명이 빠져있을 수 있는데 이럴때 수정하는 작업을 => 정제 라고 한다 (cleaning)
# 정제 작업을 포한한 정상적인 데이터 사용을 위한 포괄적인 작업을 통틀어서 전처리라고 한다
# 전처리가 정제작업보다 더 포괄적이다
# 전처리 (문자열 타입-> 날짜 타입으로 바꾼다, 성별(남,여)표기 -> 컴퓨터가 이해할 수 있게 0,1로 표기) 데이터 분석을 하기 위한 전체적인 작업
# 정제는 누가봐도 틀린 것을 바로 잡는 것
```

```python
# pasdas가 취급하는 데이터의 종류 : 2가지
# series 시리즈 : 1차원 데이터
# dataframe 데이터프레임 : 1차원 데이터가 여러개 묶여있는 형태
# 각 차원 데이터를 표기할 때, 유사 리스트의 형태로서 생성해야한다
```

```python
import pandas as pd

# seriesdata = pd.Series([70, 60, 70]) # 기본 인덱스 값으로 출력
seriesdata = pd.Series([70, 60, 70], index=["국어", "영어", "수학"]) # 인덱스에 의미를 부여
seriesdata

# 인덱스 : 리스트 등의 이터러블한 객체값을 조회 할때 인덱스 값이 생성된다
```

```python
# 설정된 인덱스의 값을 조회

seriesdata.index
```

```python
# 설정된 인덱스의 값을 수정

seriesdata.index = ["미술", "음악", "체육"]
```

```python
seriesdata.index
```

```python
# 인덱스의 값만 찾아온다

seriesdata.values
```

```python
# 해당 인덱스의 값만 찾아온다

#seriesdata["미술"]  첫번째 방법
print(seriesdata["미술"], seriesdata.iloc[0])
```

```python
seriesdata["미술"] = 80
```

```python
print(seriesdata["미술"], seriesdata.iloc[0])
```

```python
seriesdata
```

```python
# 인덱스 삭제

del seriesdata["미술"]
```

시리즈 데이터 타입 종류
- object : str | 혼용데이터 (문자열 + 정수, 정수 + 논리, 문자열 + 논리)
- int64 : 정수 (64비트)
- float64 : 실수 (64비트)
- bool : True, False
- datetime64 : 날짜, 시간 (64비트)

```python
seriesdata = pd.Series([1, 2, 5])
seriesdata
```

```python
seriesdata.astype("float")

# 이렇게하면 타입을 바꾼것 같지만, 바꾼게 아니라 실수 타입으로 보여달라고 한 것
# 바꿔서 본 걸 적용하고 싶다면 아래와 같이 써야함
```
