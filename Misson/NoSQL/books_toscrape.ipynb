- 테이블 4개를 생성 및 저장 (각 테이블 안에는 기본적으로 PK 존재해야함)

  
1) categories 테이블 (category_id, category_name)
2) books 테이블 (book_id, title, category_id)
3) prices 테이블 (price_id, book_id, price_info)
4) inventory 테이블 (inventory_id, book_id, stock_count)

1) https://books.toscrape.com/ 사이트에 접속한다
2) 메인사이트에 크롤링해와야하는 값을 확인한다
3) 책 제목, 가격, 재고여부
4) 각각의 값을 가져왔다면 수집된 데이터를 MySQL에 저장한다 pysql

```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com/"

res = requests.get(url)
soup = BeautifulSoup(res.text, "html.parser")

products = soup.select("article.product_pod")
for product in products :
    title = product.select_one("h1")
    price = product.select_one("p.price_color")
    instock = product.select_one("p.instock.availability")

    print(title, price, instock)
```
