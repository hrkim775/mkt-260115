```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By 
from webdriver_manager.chrome import ChromeDriverManager
from urllib import parse
import time


service = Service(ChromeDriverManager().install())
options = Options()

# options.add_argument("--headless") # 가상의 브라우저를 우리 눈에 보이지 않도록 설정
# options.add.argument("--desable-gpu")
options.add_argument("--disable-dev-shm-usege")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36")
options.add_argument("--lean=ko_KE")


driver = webdriver.Chrome(service=service, options = options)

target_url = "https://www.jobkorea.co.kr/Search/?stext=%EB%A7%88%EC%BC%80%ED%8C%85&Page_No=1&tabType=recruit&ord=EditDtDesc"

driver = webdriver.Chrome(service=service, options=options)
driver.get(target_url)
time.sleep(3)


job_posts = driver.find_elements(By.CSS_SELECTOR, "div.Flex_display_flex__i0l0hl2.Flex_gap_space16__i0l0hlj.Flex_direction_column__i0l0hl4")

for job_post in job_posts :
    title = job_post.find_element(By.CSS_SELECTOR, "div.styles_mb_space2__dk46ts4t").text.strip()
    #company = job_post.find_element(By.CSS_SELECTOR, "a.Flex_display_inline-flex__i0l0hl1.Flex_gap_space6__i0l0hl1g.Flex_align_center__i0l0hl8.styles_mb_space20__dk46ts2h").text.strip()
    #location = job_post.find_element(By.CSS_SELECTOR, "div.Flex_display_flex__i0l0hl2 Flex_gap_space4__i0l0hly Flex_align_center__i0l0hl8 styles_flex-shrink_0__dk46tsa9").text.strip()
    #keywords = job_post.find_element(By.CSS_SELECTOR, "div.Flex_display_flex__i0l0hl2.Flex_gap_space4__i0l0hly.Flex_align_center__i0l0hl8 styles_flex-shrink_1__dk46tsaa").text.strip()

    print(title)

driver.quit()
```
