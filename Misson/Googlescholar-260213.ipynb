1. 구글 스칼라에 접속한다
2. 구글 스칼라 사이트에서 검색창을 찾는다
3. 검색창에 마케팅이라는 키워드를 입력한다
4. 검색창에 해당 키워드가 입력되었다면 검색 버튼 또는 엔터를 누른다
5. 정상적으로 검색이 완료되면 랜딩페이지에 도착한다
6. 랜딩페이지에 도착했다면, 타이틀이 어떤 속성으로 되어있는지 개발자도구를 통해서 확인한다
7. 확인결과, 특정패턴으로 작성되어있다면 그 패턴의 선택지를 정의한다
8. 해당 선택지를 활용해서 값을 찾아온다
9. 이때 찾아온 값이 리스트 형태라는 것을 기억한다
10. 리스트는 반복문을 사용할 수 있다는 것을 기억한다
11. 해당 반복문을 통해서 값을 하나씩 찾아와서 출력한다

```python
# 1. 구글 스칼라에 접속한다

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
import csv

URL = "https://scholar.google.com/"

service = Service(ChromeDriverManager().install())
options = Options()

options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=service, options=options)
driver.get(URL)
time.sleep(2)

driver.quit()
```

```python
# 2. 구글 스칼라 사이트에서 검색창을 찾는다
# 3. 검색창에 마케팅이라는 키워드를 입력한다
# 4. 검색창에 해당 키워드가 입력되었다면 검색 버튼 또는 엔터를 누른다
# 5. 정상적으로 검색이 완료되면 랜딩페이지에 도착한다

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys # 이 구문이 있어야 엔터를 누룰 수 있다
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
import csv

URL = "https://scholar.google.com/"

service = Service(ChromeDriverManager().install())
options = Options()

options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=service, options=options)
driver.get(URL)
time.sleep(2)

search = driver.find_element(By.CSS_SELECTOR, "input.gs_in_txt.gs_in_ac") # 검색창 값을 찾아온다
search.clear() # 검색창의 값을 지운다
search.send_keys("marketing") # 검색창에 키워드를 입력한다
search.send_keys(Keys.RETURN) # 엔터를 친다
time.sleep(2)


driver.quit()
```

```python
# 6. 랜딩페이지에 도착했다면, 타이틀이 어떤 속성으로 되어있는지 개발자도구를 통해서 확인한다
# 7. 확인결과, 특정패턴으로 작성되어있다면 그 패턴의 선택지를 정의한다
# 8. 해당 선택지를 활용해서 값을 찾아온다
# 9. 이때 찾아온 값이 리스트 형태라는 것을 기억한다
# 10. 리스트는 반복문을 사용할 수 있다는 것을 기억한다
# 11. 해당 반복문을 통해서 값을 하나씩 찾아와서 출력한다

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys 
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
import csv

URL = "https://scholar.google.com/"

service = Service(ChromeDriverManager().install())
options = Options()

options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=service, options=options)
driver.get(URL)
time.sleep(2)

search = driver.find_element(By.CSS_SELECTOR, "input.gs_in_txt.gs_in_ac") 
search.clear() 
search.send_keys("marketing") 
search.send_keys(Keys.RETURN) 
time.sleep(2)

lists = driver.find_elements(By.CSS_SELECTOR, "div#gs_res_ccl_mid h3.gs_rt")

for item in lists[:5] : 
    print(item.text)


driver.quit()
```
