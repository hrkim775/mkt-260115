멜론뮤직(https://www.melon.com/index.htm) 사이트 접속 후 "소녀시대", "태연" 검색 후 출력되는 상단 아티스트 정보 크롤링 후 MongoDB > DB생성 > Collection 생성 후 저장하기

1. 메인페이지 접속
2. 검색창에 소녀시대 검색
3. 출력되는 국적, 활동유형, 활동년대, 활동장르, 데뷔곡, 결성일, 멤버 크롤링
4. 검색창에 태연 검색
5. 출력되는 국적, 활동유형, 활동년대, 활동장르,데뷔곡,생일 크롤링
6. MongoDB에 DB생성
7. Collection 생성
8. 크롤링한 데이터 저장

** 가지고 온 크롤링 데이터가 일치하지 않음 (결성일, 멤버, 생일)에 차이가 있다

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager

import re
import time

service = Service(ChromeDriverManager().install())
options = Options()

# options.add_argument("--headless")
# options.add_argument("--disable-gpu")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36")
options.add_argument("--lang=ko=KR")
options.add_argument("--no-sandbox")

target_url = "https://www.melon.com/search/total/index.htm?q=%EC%86%8C%EB%85%80%EC%8B%9C%EB%8C%80"
driver = webdriver.Chrome(service=service, options=options)
driver.get(target_url)
time.sleep(3)

atist_info = driver.find_elements(By.CSS_SELECTOR, "div.d_artist_list")

for info in atist_info :
    country = info.find_element(By.XPATH, "//dd[1]").text.strip()
    category = info.find_element(By.XPATH, "//dd[2]").text.strip()
    year = info.find_element(By.XPATH, "//dd[3]").text.strip()
    genres = info.find_element(By.CSS_SELECTOR, "dd.genre-text").text.strip()
    debutsong = info.find_element(By.CSS_SELECTOR, "dd.debut_song a.btn_play_song").text.strip()
    debutday = info.find_element(By.XPATH, "//dd[6]").text.strip()
    membernames = info.find_element(By.CSS_SELECTOR, "dd a.atistname").text.strip()
    
print(country, category, year, genres, debutsong, debutday, membernames)


driver.quit()
```
