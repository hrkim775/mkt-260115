```python
import requests
import pandas as pd

client_id = ""
client_secret = ""

url = "https://openapi.naver.com/v1/datalab/search"

headers = {
    "X-Naver-Client-Id": client_id,
    "X-Naver-Client-Secret": client_secret,
    "Content-Type": "application/json"
}

body = {
    "startDate": "2026-01-01",
    "endDate": "2026-05-13",
    "timeUnit": "date",
    "keywordGroups": [
        {"groupName": "숏폼", "keywords": ["숏폼"]},
        {"groupName": "쇼츠", "keywords": ["쇼츠"]},
        {"groupName": "릴스", "keywords": ["릴스"]},
        {"groupName": "틱톡", "keywords": ["틱톡"]},
        {"groupName": "유튜브쇼츠", "keywords": ["유튜브쇼츠"]}
    ]
}

response = requests.post(url, headers=headers, json=body)

if response.status_code == 200:
    data = response.json()

    rows = []

    for result in data["results"]:
        keyword = result["title"] if result["title"] else ""

        for item in result["data"]:
            rows.append({
                "keyword": keyword,
                "date": item.get("period", ""),
                "search": item.get("ratio", "")
            })

    df = pd.DataFrame(rows)

    # date → datetime
    df["date"] = pd.to_datetime(df["date"], errors="coerce")

    # search → float + 결측값 0 처리
    df["search"] = pd.to_numeric(df["search"], errors="coerce").fillna(0)

    # index 컬럼 추가
    df.insert(0, "index", range(1, len(df) + 1))

    print(df.head())

    df.to_csv(
        "naver_keyword_숏폼 자체 관심 증가.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("CSV 저장 완료")

else:
    print("API 요청 실패")
    print(response.status_code)
    print(response.text)
```
