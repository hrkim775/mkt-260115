```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
from urllib.parse import quote
import pandas as pd
import time

keyword = "숏폼"

start_date = "2021-04-01"
end_date = "2026-04-30"

options = Options()
options.add_argument("--headless=new")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--window-size=1920,1080")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

months = pd.date_range(start_date, end_date, freq="MS")

results = []

for d in months:
    month_start = d.strftime("%Y.%m.%d")
    month_end = (d + pd.offsets.MonthEnd(0)).strftime("%Y.%m.%d")

    print(f"{month_start} ~ {month_end} 수집 중")

    count = 0

    for page in range(1, 101):
        url = f"https://search.daum.net/search?w=news&q={quote(keyword)}&period=u&sd={month_start}&ed={month_end}&p={page}"

        driver.get(url)
        time.sleep(1.5)

        soup = BeautifulSoup(driver.page_source, "html.parser")

        articles = soup.select("a[href*='v.daum.net'], a[href*='news.v.daum.net']")

        if len(articles) == 0:
            break

        count += len(articles)

    results.append({
        "keyword": keyword,
        "month": d.strftime("%Y-%m"),
        "article_count": count
    })

driver.quit()

df = pd.DataFrame(results)

df.to_csv("daum_news_count_5years.csv", index=False, encoding="utf-8-sig")

print("완료")
print(df)
```
