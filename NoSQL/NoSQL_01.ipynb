RDBMS : 관계형 데이터베이스 관리 시스템(체계)
- 현존하는 가장 오래된 데이터 베이스 관리 시스템
- 장점 : 안정적이고 가성비가 좋고 대중적이다
- 그러나 시대가 변하면서 문제들이 생기고 불편함이 발생됨
- 정해진 패턴을 기반으로 데이터 수집, 저장하는 것이 아니라 데이터의 다양하고 매우 방대해졌다 (빅데이터 시대가 옴)
- 그렇다보니 DB를 만들고 특정 패턴을 가진 스키마-테이블을 만들어 데이터를 저장한다는게 어려워졌다
- 그래서 등장한데 NoSQL이다 -> 시대를 멈출수 없으니 시대에 맞게 문법을 수정해나간 것

NoSQL : Nit Only SQL 전통적인 SQL 문법만 의존하는 관계형데이터베이스 관리 시스템 탈피
- 고정된 스키마가 없다 -> 규칙이 없기 때문에 JOIN, 서브쿼리, 상관서브쿼리 문법자체가 불가능하다 (유사한 것은 있음)

RDBMS를 사용하는 경우
1) 최초에 한번 혹은 가끔씩만 데이터를 저장하는 경우
2) 저장된 데이터를 수정하지 않고, 그냥 읽어오는 업무 프로세스의 경우
3) MySQL, Oracle, PostgreSQL, SQLlite이 관계형 데이터 베이스 프로그램이다

NoSQL을 사용하는 경우
1) 빈번하게 데이터의 입출력이 가능해야하는 상황 (사용자의 DB값이 빈번하게 바뀌어야하는 곳 - 이커머스, 멤버십관리, 금융권쪽은 필수적으로 사용)
2) MongoDB -> 상대적으로 역사가 짧아서 다른 대체가 없음
3) JSON 형태를 띄고 있다 -> JavaScript Object Notation -> JavaScript는 웹 브라우저가 인식할 수 있는 유일한 프로그래밍 언어 그래서 Baskend 영역과 소통할 수 있다
4) 기존의 JavaScript의 문법에서 속도를 좀 더 가속화할 수 있게 만듬
5) BSON 문법 체계 -> 빅데이터를 다뤄야 하기 때문에 데이터 처리 속도가 매우 중요하기 때문에 이렇게 만들었다
