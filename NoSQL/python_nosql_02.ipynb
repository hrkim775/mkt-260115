```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client.sample_mflix
movies = db.movies
```

```python
print(movies)
```

```python
for movie in movies.find({"year": 1923}, {"_id": 0, "title": 1, "year": 1}) :
    print(movie, movie["year"])
```

```python
for movie in movies.find({"year": 1923}, {"_id": 0, "title": 1, "year": 1}) :
    print(type(movie), movie, movie["year"])
```

```python
for movie in movies.find({"year": {"$lt" : 1910}}, {"_id": 0, "title": 1, "year": 1}) :
    print(movie)
```

```python
for movie in movies.find(
    {"$or": [{"year": {"$lt": 1990}}, {"year": {"$gt": 2015}}]}, 
    {"_id": 0, "title": 1, "year": 1}
) :
    print(movie)
```

```python
for movie in movies.find(
    {"genres": {"$all": ["Action", "Sci-Fi", "Drama"]}},
    {"_id": 0, "title": 1, "year": 1, "genrees": 1}
) :
    print(movie)
```

```python
for movie in movies.find().sort("imdb.rating", -1).skip(3).limit(3) :
    print(movie)
```

```python
import re

regex = re.compile("Star", re.I)

for movie in movies.find({"title": regex}).skip(1).limit(1) :
    print(movie)
```

```python
for movie in movies.find({"title": {"$regex": "Star", "$options": "i"}}).limit(1) :
    print(movie)
```

```python
distinct_genres = movies.distinct("genres")
print(distinct_genres)
```

```python
for movie in movies.find({"genres": {"$in": ["Action", "Adenture"]}}).limit(3) :
    print(movie['genres'])
```

```python
for movie in movies.find({"writers": {"$exists": False}}).limit(3):
    print(movie)
```

```python
count = movies.count_documents({"genres": {"$in": ["Action", "Adventure"]}})
print(count)
```
