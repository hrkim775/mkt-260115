```python
client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.users

# 첫번째 방법
# result = collection.update_one({"name": "David"}, {"$set": {"age": 30}})
# print(result)

# 두번째 방법
filter = {"name": "David"}
update = {"$set": {"age": 30}} # 기존의 필드를 수정할 때는 반드시 $set을 사용해야한다
result = collection.update_one(filter, update)
print(result, result.matched_count, result.modified_count)
```

```python
client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.users

filter = {"age": {"$gt": 30}}
update = {"$set": {"is_available": True}}
result = collection.update_many(filter, update)
print("matchde Count : ", result.matched_count, "modified Count :", result.modified_count)
```

```python
client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.users

filter = {"name": "park"}
result = collection.delete_one(filter)
print("Deleted Count:", result.deleted_count)
```
