```python
# XPATH : Hyper Text Markup Language => 웹사이트를 구충, 구현할떄 필요한 마크업태그
# body, div, span, li, ul, a, p ...

# XML = Extensablr Markup Language => 사전에 정의되어있는 html태그보다 유연하게 사용할 수 있는 마크업 문법
# <system>, <spacebar> 등 내가 원하는 태그를 만들어 쓸 수 있다

# 정규화 되어있는 태그, 아이디, 클래스, 속성을 활용한 크롤링 방법을 보다 유연하게 찾아서 가져오기 위한 목적으로 탄생된 크롤링 방법 (in selenium)
# 시간이 흐르면서, xml보다는 html을 많이 쓰게 되었고, xpath보다 id, class, selector, attributes 방법으로 크롤링 하는 형태가 많아져 요즘엔 잘 쓰지 않는다

# selenium에서 크롤링을 해올 때
# 1. find_element or find_elements 중에서 선택
# 2. By.ID, By.CSS_SELECTOR, By.TAG_NAME, get_attributes()를 썼던 것 처럼
# 3. By.XPATH를 넣으면 된다

# XPATH 방식으로 특정 값을 찾아오는 문법

"""
/ : 절대경로 ->/html/bpdy/div/div
// : 특정 문서상에서 해당되는 내용을 모두 다 선책 -> //h1 이라고 썼다면, 문서상 h1태그로 시작되는 값은 다 찾아오라는 뜻
//*[@href] : 현재 문서에서 href 속성을 가지고 있는 모든 태그를 찾아라
//a[@href="https://google.com"] : 현재 문서 내 https://google.com 해당 주소값을 가지고 있는, href 속성을 가지고 있는 모든 a를 찾아와라
(//a)[3] : 현재 문서 내 모든 a 태그를 찾아와라, 그런데 3번째로 등장하는 a만 찾아와라 (주의, 0이 아니라 1부터 시작된다는 것 주의!)
(//table)[last()] : 현재 문서 내 모든 table 태그를 찾아와라, 그런데 마지막번째만 찾아와라
(//a)[position() < 3] : 현재 문서 내 모든 a 태그를 찾아와라, 그런데 3번째 보다 미만인 a만 찾아와라 (즉, 1, 2번째 a만 찾아와라)
//div[@*] : 문서상에서 div태그를 모두 찾아와라, 그런데 속성이 하나라도 존재하는 div태그를 찾아와라 (속성이 무엇이든 상관없다)
contains(@class, "속성") and contains(@class, "속성") -> 복수의 속성을 가지고 있는 경우에 한함
"""
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

URL = "https://davelee-fun.github.io/"

service = Service(ChromeDriverManager().install())
options = Options()

# options.add_argument("--headless")
# options.add_argument("--disable-gpu")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=service, options=options)
driver.get(URL)

# element = driver.find_element(By.XPATH, "/html/body/div[1]/div[1]/div[1]/h1")
element = driver.find_element(By.XPATH, "//h1") # 구조가 잘 변하지 않기 때문에 이게 더 안정적일때도 있다
elements = driver.find_elements(By.XPATH, "//*[@href]") 

# print(element.text)

for element in elements :
    print(element.text)

driver.quit()
```
